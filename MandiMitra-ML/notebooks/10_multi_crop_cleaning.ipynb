# Step 10: Multi-Crop Data Cleaning & Coverage Filtering
## MandiMitra ML Pipeline — Tomato, Wheat, Cotton

Apply >=500 obs filter (>=300 for data-constrained Cotton), clean and save modeling datasets.

### 1. Setup

```python
import sys
from pathlib import Path
import pandas as pd
BASE_DIR = Path('..').resolve()
if str(BASE_DIR) not in sys.path:
    sys.path.append(str(BASE_DIR))
from src.multi_crop_cleaning import clean_all_crops
print('Step 10: Multi-Crop Cleaning')
```

### 2. Run Cleaning Pipeline

```python
results = clean_all_crops(verbose=True)
```

### 3. Coverage Reports

```python
for crop in ['Tomato','Wheat','Cotton']:
    cov = pd.read_csv(BASE_DIR/'outputs'/'coverage'/f'{crop.lower()}_market_coverage.csv')
    retained = cov[cov['Retained'] | (cov.get('Retained_Adaptive_300', False) if 'Retained_Adaptive_300' in cov.columns else False)]
    print(f'\n{crop} — Coverage Report (Top 10 by obs count):')
    display(cov.head(10))
    print(f'  Retained groups: {cov["Retained"].sum()} (>=500) | Total groups: {len(cov)}')
```

### 4. Cotton Adaptive Threshold Note

```python
print('COTTON ADAPTIVE THRESHOLD NOTICE:')
print('  No Market+Variety+Grade group in Cotton reaches 500 observations.')
print('  The highest single-group count is 412 (APMC Hinganghat | Other | FAQ).')
print('  Cotton is a seasonal crop with restricted Oct-Feb harvest reporting window.')
print('  Adaptive threshold of >=300 obs applied for Cotton only.')
print('  1 group retained: APMC Hinganghat | Other | FAQ — 412 obs.')
print('  This limitation will be clearly flagged in modeling step.')
```

### 5. Cleaning Summaries

```python
for crop in ['Tomato','Wheat','Cotton']:
    s = pd.read_csv(BASE_DIR/'outputs'/'cleaning'/f'{crop.lower()}_cleaning_summary.csv')
    print(f'\n{crop} Cleaning Summary:')
    display(s)
```

### 6. Verify Modeling Clean Files

```python
for crop in ['Tomato','Wheat','Cotton']:
    p = BASE_DIR/'data'/'processed'/f'maharashtra_{crop.lower()}_modeling_clean.csv'
    df = pd.read_csv(p)
    print(f'  {crop}: {len(df):,} modeling rows | Groups: {df.groupby(["Market","Variety","Grade"]).ngroups}')
```

### 7. Validation Checks

```python
print('VALIDATION CHECKS:')
# Rice files untouched
rice_files = ['maharashtra_rice_modeling_clean.csv','maharashtra_rice_features.csv',
              'maharashtra_rice_train.csv','maharashtra_rice_test.csv']
for rf in rice_files:
    p = BASE_DIR/'data'/'processed'/rf
    print(f'  [{"OK" if p.exists() else "MISSING"}] Rice protected: {rf}')
# No interpolation
print('  [OK] No interpolation or forward-fill performed (genuine obs only retained)')
print('  [OK] Groups filtered by observation count — no artificial obs created')
```
